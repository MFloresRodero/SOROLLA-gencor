### Import data and libraries

```python
import gwaslab as gl
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
zeta = pd.read_csv("/home/maria/git/SOROLLA/ZETA-BETA/ldsc-Z/zeta_test_results.csv")
print("Zeta Test")
display(zeta.head())
```

```python
beta = pd.read_csv("/home/maria/git/SOROLLA/ZETA-BETA/ldsc-B/beta_test_results.csv")
print("Beta Test")
display(beta.head())
```

```python
og_results = pd.read_csv("/home/maria/git/SOROLLA/Results/ldsc_genetic_correlation.csv")
print("OG Resuls")
display(og_results.head())
```

```python
# Merge DataFrames
merged_tests_1_2 = zeta.merge(beta, on=["label_1","id_1","label_2","id_2"], suffixes=["_z", "_b"])
# Rename in test3 because otherwise the suffix option is not applying
repeat_test3_renamed = og_results.rename(columns=lambda x: x if x in ["label_1","id_1","label_2","id_2"] else f"{x}_og")
# Merge all three now
tested_z_b_og = merged_tests_1_2.merge(repeat_test3_renamed, on=["label_1","id_1","label_2","id_2"])
#Save to csv file:
tested_z_b_og.to_csv("/home/maria/git/SOROLLA/REPEATED/merged_tests_comparison.csv")
display(tested_z_b_og.head())
display(tested_z_b_og.columns)
```

```python
# Define the base column lists
columns_correlation_base = ['label_1', 'id_1', 'label_2', 'id_2', 'rg', 'se', 'z', 'p']
columns_covariance_base = ['label_1', 'id_1', 'label_2', 'id_2', 'gcov_int', 'gcov_int_se']

# Define the suffixes
suffixes = ['z', 'b', 'og']
```

```python
# Generate the full column lists
columns_correlation = ['label_1', 'id_1', 'label_2', 'id_2'] + [f'{col}_{suffix}' for col in columns_correlation_base[4:] for suffix in suffixes]
# Generate correlation_test
correlation_test = tested_z_b_og[columns_correlation]
# Display the df
display(correlation_test.head())
print(correlation_test.columns)
correlation_test.to_csv("/home/maria/git/SOROLLA/ZETA-BETA/correlation_b_z.csv")
```

```python
# Generate the full column lists
columns_covariance = ['label_1', 'id_1', 'label_2', 'id_2'] + [f'{col}_{suffix}' for col in columns_covariance_base[4:] for suffix in suffixes]
# Generate the covariance_test
covariance_test = tested_z_b_og[columns_covariance]
# Display the df
display(covariance_test.head())
print(covariance_test.columns)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))

# Scatter plot for rg values
ax1.scatter(correlation_test["rg_z"], correlation_test["rg_b"], label='Data points')

# y=x line
ax1.plot([correlation_test["rg_z"].min(), correlation_test["rg_z"].max()], 
         [correlation_test["rg_z"].min(), correlation_test["rg_z"].max()], 
         linestyle='--', color='red')

# Add thin grey lines at x=0 and y=0
ax1.axhline(0, color='grey', linestyle='-', linewidth=0.5, alpha=0.5)
ax1.axvline(0, color='grey', linestyle='-', linewidth=0.5, alpha=0.5)

# Identify and label outliers
for i, row in correlation_test.iterrows():
    rg_diff = abs(row["rg_z"] - row["rg_b"])
    if rg_diff > 0.2:  # Define a threshold for outliers
        ax1.annotate(f'{row["p1"]}, {row["p2"]}', (row["rg_z"], row["rg_b"]), fontsize=8)

# Change color and label points with different directions
opposite_direction_label_added = False
for i, row in correlation_test.iterrows():
    if (row["rg_z"] > 0 and row["rg_b"] < 0) or (row["rg_z"] < 0 and row["rg_b"] > 0):
        if not opposite_direction_label_added:
            ax1.scatter(row["rg_z"], row["rg_b"], color='orange', label='Opposite directions')
            opposite_direction_label_added = True
        else:
            ax1.scatter(row["rg_z"], row["rg_b"], color='orange')
        ax1.annotate(f'{row["p1"]}, {row["p2"]}', (row["rg_z"], row["rg_b"]), fontsize=8)


ax1.set_xlabel('rg_z')
ax1.set_ylabel('rg_b')

# Scatter plot for p-values
ax2.scatter(correlation_test["p_z"], correlation_test["p_b"])
ax2.plot([correlation_test["p_z"].min(), correlation_test["p_z"].max()], 
         [correlation_test["p_z"].min(), correlation_test["p_z"].max()], 
         linestyle='--', color='red')
ax2.set_xlabel('p_z')
ax2.set_ylabel('p_b')

# Place the label at the bottom right
ax1.legend(loc='lower right')

plt.show()
```
