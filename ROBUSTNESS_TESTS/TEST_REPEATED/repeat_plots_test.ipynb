```python
repeat_test3 = pd.read_csv("/home/maria/git/SOROLLA/REPEATED/ldsc_test3/ldsc_correlation_test3.csv")
print("Test repeat 3")
display(repeat_test3.head())
```

```python
# Merge DataFrames
merged_tests_1_2 = repeat_test1.merge(repeat_test2, on=["label_1","id_1","label_2","id_2"], suffixes=["_test1", "_test2"])
# Rename in test3 because otherwise the suffix option is not applying
repeat_test3_renamed = repeat_test3.rename(columns=lambda x: x if x in ["label_1","id_1","label_2","id_2"] else f"{x}_test3")
# Merge all three now
merged_tests = merged_tests_1_2.merge(repeat_test3_renamed, on=["label_1","id_1","label_2","id_2"])
#Save to csv file:
merged_tests.to_csv("/home/maria/git/SOROLLA/REPEATED/merged_tests_comparison.csv")
display(merged_tests.head())
display(merged_tests.columns)
```

```python
# Define the base column lists
columns_correlation_base = ['label_1', 'id_1', 'label_2', 'id_2', 'rg', 'se', 'z', 'p']
columns_covariance_base = ['label_1', 'id_1', 'label_2', 'id_2', 'gcov_int', 'gcov_int_se']

# Define the suffixes
suffixes = ['test1', 'test2', 'test3']
```

```python
# Generate the full column lists
columns_correlation = ['label_1', 'id_1', 'label_2', 'id_2'] + [f'{col}_{suffix}' for col in columns_correlation_base[4:] for suffix in suffixes]
# Generate correlation_test
correlation_test = merged_tests[columns_correlation]
# Display the df
display(correlation_test.head())
print(correlation_test.columns)
```

```python
# Generate the full column lists
columns_covariance = ['label_1', 'id_1', 'label_2', 'id_2'] + [f'{col}_{suffix}' for col in columns_covariance_base[4:] for suffix in suffixes]
# Generate the covariance_test
covariance_test = merged_tests[columns_covariance]
# Display the df
display(covariance_test.head())
print(covariance_test.columns)
```
